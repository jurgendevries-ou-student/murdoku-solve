```python
import clingo

def load_program(filename):
    ctl = clingo.Control(["0"])
    ctl.load(filename + '.lp')
    ctl.ground([("base", [])])
    return ctl
    
def solve_program(filename, count=False):
    ctl = load_program(filename)
    def on_model(model):
        print(model.symbols(shown=True))
    with ctl.solve(on_model=on_model, yield_=True) as handle:
        models = list(handle)
        result = handle.get()
    if count:
        print(f'Number of models: {len(models)}')
        
    print(result)
    print("Satisfiable:", result.satisfiable)
    print("Unsatisfiable:", result.unsatisfiable)
    print("Unknown:", result.unknown)
    return result
```

```python
solve_program('program3')
```
